```python
import pandas as pd

# Load the data from TSV files
languages_df = pd.read_csv('languages.tsv', sep='\t')
forms_df = pd.read_csv('forms.tsv', sep='\t')

# Display basic information about both dataframes
print("Languages DataFrame Info:")
print(languages_df.info())
print("\nForms DataFrame Info:")
print(forms_df.info())

# Display first few rows of both dataframes
print("\nFirst few rows of Languages DataFrame:")
print(languages_df.head())
print("\nFirst few rows of Forms DataFrame:")
print(forms_df.head())
```

```python
forms_df.columns = ['isocode', 'concept', 'form', 'wordLength', 'longestClusterLength', 'vowConsRatio']
print(forms_df.head())
```

```python
# Select only the relevant columns from languages_df
num_cols = ['inventorySize', 'phonVowLength', 'phonGeminates']
id_col = 'LanguageID'  # Assuming 'LanguageID' is equivalent to 'isocode' in your context

for col in num_cols:
    min_idx = languages_df[col].idxmin()
    max_idx = languages_df[col].idxmax()
    min_lang = languages_df.loc[min_idx, id_col]
    max_lang = languages_df.loc[max_idx, id_col]
    min_val = languages_df.loc[min_idx, col]
    max_val = languages_df.loc[max_idx, col]
    print(f"For {col}:")
    print(f"  Min: {min_val} ({min_lang})")
    print(f"  Max: {max_val} ({max_lang})\n")
```
