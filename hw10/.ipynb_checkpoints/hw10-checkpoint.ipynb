```python
import pandas as pd
import numpy as np
from collections import Counter, defaultdict
import re
import math
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import AgglomerativeClustering, KMeans
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import dendrogram
from scipy.spatial.distance import pdist, squareform
import warnings
warnings.filterwarnings('ignore')
```

```python
# Task 1a: Load and tokenize the corpus
def load_and_tokenize_corpus(filename):
    tokens = []
    
    with open(filename, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:  # Skip empty lines
                continue
            
            # Split into sentences by sentence-final marks
            sentences = re.split(r'[!?.]+', line)
            
            for sentence in sentences:
                sentence = sentence.strip()
                if not sentence:
                    continue
                
                # Split into tokens by spaces
                words = sentence.split()
                
                for word in words:
                    # Remove trailing punctuation
                    token = re.sub(r'[,.:]+$', '', word)
                    # Convert to lowercase
                    token = token.lower()
                    # Add if not empty
                    if token:
                        tokens.append(token)
    
    return tokens

# Load the corpus
tokens = load_and_tokenize_corpus('cym-os-corpus.txt')
print(f"Total tokens: {len(tokens)}")
print(f"First 20 tokens: {tokens[:20]}")
```

```python
# Task 1b: Character frequency analysis
def analyze_characters(tokens):
    """
    Count all characters and compute log probabilities
    """
    char_counter = Counter()
    
    # Count all characters in all tokens
    for token in tokens:
        for char in token:
            char_counter[char] += 1
    
    # Sort by frequency
    sorted_chars = char_counter.most_common()
    
    # Calculate total characters
    total_chars = sum(char_counter.values())
    
    # Compute log probabilities
    log_char_prob = {}
    for char, count in char_counter.items():
        probability = count / total_chars
        log_char_prob[char] = math.log(probability)
    
    return sorted_chars, log_char_prob

print("\nAnalyzing character frequencies...")
sorted_chars, log_char_prob = analyze_characters(tokens)

print(f"Most frequent characters:")
for char, count in sorted_chars[:10]:
    print(f"'{char}': {count:,}")

print(f"\nTotal unique characters: {len(sorted_chars)}")
print(f"Most frequent character '{sorted_chars[0][0]}' occurs {sorted_chars[0][1]:,} times")
```

```python
# Task 1c: Token frequency analysis
def analyze_tokens(tokens):
    """
    Count token frequencies and extract most common
    """
    token_counter = Counter(tokens)
    most_common_tokens = token_counter.most_common(20)
    
    return token_counter, most_common_tokens

print("\nAnalyzing token frequencies...")
token_counter, most_common_tokens = analyze_tokens(tokens)

print(f"20 most common tokens:")
for i, (token, count) in enumerate(most_common_tokens, 1):
    print(f"{i:2d}. '{token}': {count:,}")

print(f"\nTotal unique tokens: {len(token_counter)}")
print(f"Most frequent token '{most_common_tokens[0][0]}' occurs {most_common_tokens[0][1]:,} times")

# Store data for later tasks
print(f"\nData summary:")
print(f"- Total tokens: {len(tokens):,}")
print(f"- Unique tokens: {len(token_counter):,}")
print(f"- Unique characters: {len(sorted_chars)}")
print(f"- Character log probabilities stored in log_char_prob dictionary")
```

```python
# Task 2: Identifying Vowels and Consonants through Clustering

# Task 2a: Build dictionary of character environments
def build_character_environments(tokens):
    """
    Build a dictionary storing environments for each character.
    Environments are of shape A_B where A and B are preceding/following characters.
    Pad tokens with # at start and end.
    """
    char_environments = defaultdict(set)
    
    for token in tokens:
        # Pad token with sentinel symbols
        padded_token = '#' + token + '#'
        
        # Extract environments for each character
        for i in range(1, len(padded_token) - 1):  # Skip the padding characters themselves
            char = padded_token[i]
            preceding = padded_token[i-1]
            following = padded_token[i+1]
            
            # Environment format: A_B
            environment = f"{preceding}_{following}"
            char_environments[char].add(environment)
    
    return char_environments

print("Building character environments...")
char_environments = build_character_environments(tokens)

print(f"Number of characters with environments: {len(char_environments)}")
print(f"Example - environments for 'y': {list(char_environments['y'])[:10]}")
print(f"Example - environments for 'd': {list(char_environments['d'])[:10]}")
```

```python
# Task 2b: Convert to DataFrame with binary features
def create_environment_dataframe(char_environments, min_environments=200):
    """
    Convert character environments to binary DataFrame.
    Filter to characters with at least min_environments different environments.
    """
    # Filter characters by minimum environment count
    filtered_chars = {char: envs for char, envs in char_environments.items() 
                     if len(envs) >= min_environments}
    
    print(f"Characters with at least {min_environments} environments: {len(filtered_chars)}")
    
    # Get all unique environments
    all_environments = set()
    for envs in filtered_chars.values():
        all_environments.update(envs)
    
    all_environments = sorted(all_environments)
    
    # Create DataFrame
    data = []
    for char in sorted(filtered_chars.keys()):
        row = [1 if env in filtered_chars[char] else 0 for env in all_environments]
        data.append(row)
    
    df = pd.DataFrame(data, 
                     index=sorted(filtered_chars.keys()), 
                     columns=all_environments)
    
    return df

print("\nCreating environment DataFrame...")
env_df = create_environment_dataframe(char_environments)

print(f"DataFrame shape: {env_df.shape}")
print(f"Characters (rows): {env_df.shape[0]}")
print(f"Environments (columns): {env_df.shape[1]}")
print(f"Sample of the DataFrame:")
print(env_df.iloc[:5, :10])
```

```python
# Task 2c: Agglomerative clustering and dendrogram
def cluster_characters_by_environment(df):
    """
    Perform agglomerative clustering on characters based on their environments.
    Create dendrogram visualization.
    """
    # Perform agglomerative clustering
    clustering = AgglomerativeClustering(n_clusters=None, 
                                       distance_threshold=0,
                                       linkage='ward')
    
    # Fit the clustering
    clustering.fit(df.values)
    
    # Create dendrogram (adapted from scikit-learn documentation)
    def plot_dendrogram(model, **kwargs):
        # Create linkage matrix
        counts = np.zeros(model.children_.shape[0])
        n_samples = len(model.labels_)
        for i, merge in enumerate(model.children_):
            current_count = 0
            for child_idx in merge:
                if child_idx < n_samples:
                    current_count += 1
                else:
                    current_count += counts[child_idx - n_samples]
            counts[i] = current_count

        linkage_matrix = np.column_stack([model.children_, model.distances_,
                                        counts]).astype(float)

        # Plot the dendrogram
        plt.figure(figsize=(15, 8))
        dendrogram(linkage_matrix, labels=df.index, leaf_rotation=90, **kwargs)
        plt.title('Character Clustering Dendrogram\n(Based on Environmental Similarity)')
        plt.xlabel('Characters')
        plt.ylabel('Distance')
        plt.tight_layout()
        plt.show()

    plot_dendrogram(clustering)
    
    return clustering

print("\nPerforming agglomerative clustering...")
clustering_result = cluster_characters_by_environment(env_df)

# Analyze the clustering results
def analyze_clustering_results(df, clustering_result, n_clusters=5):
    """
    Analyze clustering results by extracting clusters and examining them.
    """
    # Get cluster assignments for a specific number of clusters
    clustering_specific = AgglomerativeClustering(n_clusters=n_clusters, linkage='ward')
    cluster_labels = clustering_specific.fit_predict(df.values)
    
    # Group characters by cluster
    clusters = defaultdict(list)
    for char, cluster_id in zip(df.index, cluster_labels):
        clusters[cluster_id].append(char)
    
    print(f"\nClustering results with {n_clusters} clusters:")
    for cluster_id, chars in clusters.items():
        print(f"Cluster {cluster_id}: {sorted(chars)}")
    
    # Identify potential vowels and consonants
    print(f"\nAnalysis of clusters:")
    print("Expected English vowels: a, e, i, o, u")
    print("Characters that might be vowels in Welsh based on clustering:")
    
    for cluster_id, chars in clusters.items():
        # Look for clusters containing known vowels
        known_vowels = set(['a', 'e', 'i', 'o', 'u'])
        cluster_set = set(chars)
        vowel_overlap = known_vowels.intersection(cluster_set)
        
        if vowel_overlap:
            print(f"  Cluster {cluster_id} (contains known vowels {vowel_overlap}): {sorted(chars)}")
            potential_new_vowels = cluster_set - known_vowels
            if potential_new_vowels:
                print(f"    -> Potential additional vowels: {sorted(potential_new_vowels)}")
    
    return clusters

clusters = analyze_clustering_results(env_df, clustering_result)
```

Conclusion:
    Based on the clustering of characters by their environments, we can identify which characters behave similarly in Welsh.
    Characters that cluster with known vowels (a, e, i, o, u) are likely to function as vowels in Welsh orthography, even if they represent consonants in English (such as 'w' and 'y').

This explains how words like 'yw', 'ydyw', 'fyddwn' can be pronounced - some of these 'consonant' symbols actually function as vowels in Welsh!

```python
# Task 3a: Search for Potential Digraphs

# Define consonants from Task 2a clustering results
CONSONANTS = set(["'", 'b', 'c', 'd', 'f', 'g', 'h', 'k', 'l', 'm', 'n', 'p', 'r', 's', 't'])

def extract_candidate_digraphs(tokens, consonants):
    """
    Extract candidate digraphs from tokens.
    Two types: geminates (identical consonants) and consonant + h combinations.
    Count total occurrences and word-internal occurrences.
    Process each word form only once (ignore token frequency).
    """
    
    # Use set to process each word form only once
    unique_tokens = set(tokens)
    
    # Counters for candidate digraphs
    cons_bigram_counter = Counter()
    internal_bigram_counter = Counter()
    
    print(f"Processing {len(unique_tokens)} unique word forms...")
    
    for token in unique_tokens:
        token_length = len(token)
        
        # Look for bigrams in the token
        for i in range(token_length - 1):
            char1 = token[i]
            char2 = token[i + 1]
            
            # Check if this is a candidate digraph
            is_candidate = False
            bigram = char1 + char2
            
            # Type 1: Geminates (identical consonants)
            if char1 == char2 and char1 in consonants:
                is_candidate = True
            
            # Type 2: Consonant + h
            elif char1 in consonants and char2 == 'h':
                is_candidate = True
            
            if is_candidate:
                # Count total occurrence
                cons_bigram_counter[bigram] += 1
                
                # Check if it's word-internal (not at start or end)
                if i > 0 and i < token_length - 2:
                    internal_bigram_counter[bigram] += 1
    
    return cons_bigram_counter, internal_bigram_counter
```

```python

# Extract candidate digraphs
print("Task 3a: Extracting candidate digraphs...")
print(f"Consonants: {sorted(CONSONANTS)}")

cons_bigram_counter, internal_bigram_counter = extract_candidate_digraphs(tokens, CONSONANTS)

print(f"\nFound {len(cons_bigram_counter)} candidate digraphs")
print(f"Most common candidate digraphs:")
for bigram, count in cons_bigram_counter.most_common(10):
    internal_count = internal_bigram_counter.get(bigram, 0)
    internal_ratio = internal_count / count if count > 0 else 0
    print(f"  '{bigram}': {count} total, {internal_count} internal ({internal_ratio:.3f} ratio)")

```

```python
# Analyze the types of candidates found
print(f"\nAnalysis of candidate types:")

geminates = []
cons_h_combinations = []

for bigram in cons_bigram_counter:
    if len(bigram) == 2:
        char1, char2 = bigram[0], bigram[1]
        if char1 == char2:
            geminates.append(bigram)
        elif char2 == 'h':
            cons_h_combinations.append(bigram)

print(f"Geminates found: {sorted(geminates)}")
print(f"Consonant + h combinations found: {sorted(cons_h_combinations)}")


```

Insights for Task 3b:
- Good digraphs should have HIGH frequency (high total_count)
- Good digraphs should have LOW internal_ratio (appear at word boundaries)
- Task 3b will use these two dimensions for clustering

```python
# Create summary of results
def create_digraph_summary(cons_bigram_counter, internal_bigram_counter):
    """
    Create a summary of digraph candidates with their statistics
    """
    summary_data = []
    
    for bigram, total_count in cons_bigram_counter.items():
        internal_count = internal_bigram_counter.get(bigram, 0)
        internal_ratio = internal_count / total_count if total_count > 0 else 0
        
        # Classify type
        if len(bigram) == 2:
            char1, char2 = bigram[0], bigram[1]
            if char1 == char2:
                bigram_type = "geminate"
            elif char2 == 'h':
                bigram_type = "consonant_h"
            else:
                bigram_type = "other"
        else:
            bigram_type = "other"
        
        summary_data.append({
            'bigram': bigram,
            'type': bigram_type,
            'total_count': total_count,
            'internal_count': internal_count,
            'internal_ratio': internal_ratio
        })
    
    return sorted(summary_data, key=lambda x: x['total_count'], reverse=True)
```

```python
# Export results for Task 3b
def export_task3a_results(summary_data, filename="task3a_digraph_candidates.tsv"):
    """
    Export candidate digraphs for Task 3b analysis
    """
    df = pd.DataFrame(summary_data)
    df.to_csv(filename, sep='\t', index=False)
    print(f"\nExported {len(summary_data)} candidate digraphs to {filename}")
    return df

# Export results
digraph_df = export_task3a_results(summary_data)
```
