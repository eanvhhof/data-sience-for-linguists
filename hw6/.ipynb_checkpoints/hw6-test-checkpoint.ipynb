```python
import pandas as pd

# Load the CSV files into DataFrames
languages = pd.read_csv('languages.csv')
parameters = pd.read_csv('parameters.csv')
values = pd.read_csv('values.csv')
codes = pd.read_csv('codes.csv')

# Display basic information about the loaded DataFrames
print("Loaded DataFrames shapes:")
print(f"languages: {languages.shape}")
print(f"parameters: {parameters.shape}")
print(f"values: {values.shape}")
print(f"codes: {codes.shape}")
```

```python
# Filter values DataFrame for specific parameters and remove unnecessary columns
values_filtered = values[values['Parameter_ID'].isin(['GB020', 'GB021', 'GB022'])].drop(
    columns=['ID', 'Comment', 'Source', 'Source_comment', 'Coders']
)
print("Shape of filtered values DataFrame:", values_filtered.shape)
print("\nFirst few rows of filtered values DataFrame:")
print(values_filtered.head())
```

```python
# Merge languages and values_filtered dataframes
merged_df = values_filtered.merge(
    languages[['ID', 'Name', 'Macroarea']], 
    left_on='Language_ID', 
    right_on='ID',
    how='left'
).drop('ID', axis=1)

print("Shape of merged DataFrame:", merged_df.shape)
print("\nFirst few rows of merged DataFrame:")
print(merged_df.head())
```

```python
final_df = merged_df.merge(
    codes[['ID', 'Description']], 
    left_on='Code_ID', 
    right_on='ID',
    how='left'
).rename(columns={'ID': 'ID_y'})

print("Shape of final DataFrame:", final_df.shape)
print("\nLast three rows of final DataFrame:")
print(final_df.tail(3))
```

```python
# Remove unnecessary columns and rename the remaining ones
cleaned_df = final_df.drop(['Language_ID', 'Code_ID', 'ID_y'], axis=1).rename(
    columns={
        'Name': 'Language',
        'Parameter_ID': 'Feature',
        'Description': 'Value'
    }
)

# Display the results
print("Shape of cleaned DataFrame:", cleaned_df.shape)
print("\nFirst few rows of cleaned DataFrame:")
print(cleaned_df.head())
```

```python
# Reorder columns and sort by Macroarea and Language
cleaned_df = cleaned_df[['Language', 'Macroarea', 'Feature', 'Value']].sort_values(
    by=['Macroarea', 'Language']
).reset_index(drop=True)

print("First few rows of reordered and sorted DataFrame:")
print(cleaned_df.head())
```

```python
# Remove rows with missing values
cleaned_df = cleaned_df.dropna()

# Create a mapping dictionary for feature names
feature_mapping = {
    'GB020': 'defArt',
    'GB021': 'prenom',
    'GB022': 'postnom'
}

# Replace feature names
cleaned_df['Feature'] = cleaned_df['Feature'].replace(feature_mapping)

# Reset index to ensure continuous numbering
cleaned_df = cleaned_df.reset_index(drop=True)

print("Last three rows of the cleaned DataFrame:")
print(cleaned_df.tail(3))
```

```python
# Create hierarchical index using set_index
hierarchical_df = cleaned_df.set_index(['Macroarea', 'Language'])

# Display the result
print("First few rows of the hierarchical DataFrame:")
print(hierarchical_df.head())
```

```python
# Calculate global distribution
# Calculate global distribution
defArt_data = cleaned_df[cleaned_df['Feature'] == 'defArt']
global_dist = defArt_data.loc[:, 'Value'].value_counts()
print("Global distribution of definite articles:")
print(global_dist)
print("\nPercentage:")
print((global_dist / len(defArt_data) * 100).round(2))

# Calculate distribution for Eurasia
eurasia_data = defArt_data[defArt_data['Macroarea'] == 'Eurasia']
eurasia_dist = eurasia_data.loc[:, 'Value'].value_counts()
print("\nEurasia distribution:")
print(eurasia_dist)
print("\nPercentage:")
print((eurasia_dist / len(eurasia_data) * 100).round(2))

# Calculate distribution for Africa
africa_data = defArt_data[defArt_data['Macroarea'] == 'Africa']
africa_dist = africa_data.loc[:, 'Value'].value_counts()
print("\nAfrica distribution:")
print(africa_dist)
print("\nPercentage:")
print((africa_dist / len(africa_data) * 100).round(2))
```

```python
# Create new hierarchical index with Feature as outer level and Macroarea as inner level
reindexed = cleaned_df.set_index(['Feature', 'Macroarea', 'Language']).sort_index()

# Display the result
print("First few rows of the reindexed DataFrame:")
print(reindexed.head())
```

```python
# Reindex with Macroarea as outer level and Feature as inner level
reindexed = cleaned_df.set_index(['Macroarea', 'Feature', 'Language']).sort_index()

# Cross-section to get languages with prenominal articles
prenominal = reindexed.xs('prenom', level='Feature')
prenominal = prenominal[prenominal['Value'] == 'Yes']
print("Languages with prenominal articles by macroarea:")
print(prenominal)
```
